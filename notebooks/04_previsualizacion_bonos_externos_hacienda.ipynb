```python
import pandas as pd

url = "https://www.hacienda.cl/areas-de-trabajo/finanzas-internacionales/oficina-de-la-deuda-publica/estadisticas/caracteristicas-financieras-bonos-externos"

tablas = pd.read_html(url)
print(f"Se encontraron {len(tablas)} tablas en la página")
print(tablas[0])
```

```python
def limpiar_tabla(tabla: pd.DataFrame) -> pd.DataFrame:
    """Voltea la tabla: de 'campos como filas, bonos como columnas'
    a 'bonos como filas, campos como columnas'."""
    tabla = tabla.set_index("Emisiones")
    return tabla.transpose()

tablas_limpias = [limpiar_tabla(t) for t in tablas]
df_bonos = pd.concat(tablas_limpias)
df_bonos.index.name = "bono"
df_bonos = df_bonos.reset_index()

print(f"Total de bonos: {len(df_bonos)}")
print(df_bonos.columns.tolist())
df_bonos.head(10)
```

```python
import re

def normalizar_nombre_columna(nombre: str) -> str:
    """Colapsa espacios múltiples a uno solo y quita espacios al inicio/final."""
    return re.sub(r"\s+", " ", nombre).strip()

df_bonos.columns = [normalizar_nombre_columna(c) for c in df_bonos.columns]
print(df_bonos.columns.tolist())
```

```python
print(df_bonos["bono"].value_counts().head(10))
```

```python
duplicados_exactos = df_bonos[df_bonos.duplicated(keep=False)].sort_values("bono")
print(duplicados_exactos[["bono", "Fecha emisión", "Fecha vencimiento"]])
```

```python
print(f"Filas antes: {len(df_bonos)}")
df_bonos = df_bonos.drop_duplicates()
print(f"Filas después: {len(df_bonos)}")
```

```python
df_bonos["bono"] = df_bonos["bono"].str.replace(r"\.\d+$", "", regex=True)
print(df_bonos[["bono", "Fecha emisión", "Fecha vencimiento"]].tail(15))
```

```python
df_bonos["monto_raw"] = df_bonos["Monto (MM US$)(1)"].fillna(df_bonos["Monto (MM)"])
print(df_bonos[["bono", "Monto (MM US$)(1)", "Monto (MM)", "monto_raw"]].head(35))
```

```python
print(df_bonos["monto_raw"].isna().sum())
```

```python
df_bonos = df_bonos.drop(columns=["Monto (MM US$)(1)", "Monto (MM)"])
```
